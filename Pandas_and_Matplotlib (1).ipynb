```python
data_frame = data_frame.dropna()
data_frame
```

```python
df = data_frame.drop_duplicates()
df
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(df["Date"], df["Sales"])
plt.xlabel("Date")
plt.ylabel("Sales xyz")
plt.title("Sales Over Time")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(df["Category"], df["Sales"])
plt.ylabel("Price")
plt.xlabel("Medicine")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.scatter(data_frame["Price"], data_frame["Sales"])
plt.xlabel("Price")
plt.show()

plt.hist(df["Medicine"])
plt.show()
```

```python
summary = df.groupby("Medicine")["Sales"].sum()

summary.plot(kind="bar")
plt.title("Total Sales by Medicine")
plt.xlabel("Medicine")
plt.ylabel("Sales")
plt.show()
```

```python
import seaborn as sns
```

```python
sns.histplot(data=df, x="Price")
plt.title("Distribution of Medicine Prices")
plt.xlabel("Price")
plt.ylabel("Number of Medicines")
plt.show()
```

```python
sns.countplot(data=df, x="Category")
plt.title("Number of Medicines by Category")
plt.xlabel("Medicine Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(data=df, x="Category", y="Price")
plt.title("Average Medicine Price by Category")
plt.xlabel("Category")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(data=df, x="Category", y="Price")
plt.title("Medicine Price Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(data=df, x="Category", y="Sales")
plt.title("Average Sales by Category")
plt.tight_layout()
plt.savefig("sales_by_category.png")
plt.show()
```
